```python
#!/usr/bin/env python3
"""
Minimal helper for Jupyter notebooks: read a CSV column and return its values as a Python list.

Usage in a notebook:
  from csv_column_to_array_notebook import read_csv_column
  values = read_csv_column("data.csv", "name")   # or column index 0
  values[:10]  # inspect first 10 entries
"""
import csv
from typing import List, Union

def read_csv_column(path: str, column: Union[int, str], delimiter: str = ',', skip_header: bool = False, skip_empty: bool = False) -> List[str]:
    """
    Read a single column from a CSV and return a list of values.

    - path: path to CSV file
    - column: column name (str) or zero-based index (int)
    - delimiter: CSV delimiter (default ',')
    - skip_header: if True and `column` is an int, skip the first row (useful when file has header)
    - skip_empty: if True, skip empty strings

    Returns a list of strings (empty cells become '').
    """
    values: List[str] = []
    # detect numeric index vs name
    is_index = isinstance(column, int)
    with open(path, newline='', encoding='utf-8') as f:
        if is_index:
            reader = csv.reader(f, delimiter=delimiter)
            for i, row in enumerate(reader):
                if i == 0 and skip_header:
                    continue
                val = row[column] if column < len(row) else ''
                if skip_empty and val == '':
                    continue
                values.append(val)
        else:
            reader = csv.DictReader(f, delimiter=delimiter)
            if reader.fieldnames is None or column not in reader.fieldnames:
                raise ValueError(f"Column name '{column}' not found. Available: {reader.fieldnames}")
            for row in reader:
                val = row.get(column, '')
                if skip_empty and val == '':
                    continue
                values.append(val)
    return values

# Example quick-run when imported into a notebook:
# if __name__ == "__main__":
#     vals = read_csv_column("example.csv", "name")
#     print(vals)
```
